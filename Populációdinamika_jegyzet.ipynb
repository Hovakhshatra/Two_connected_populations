#Populációdinamika

A populációdinamika a biológia és a matematika határterületén elhelyezkedő tudományág, amely a populációk méretének és szerkezetének időbeli és térbeli változásait vizsgálja. A populációk egy adott faj egyedeinek csoportjai, amelyek egy meghatározott földrajzi területen élnek, és egymással kölcsönhatásban állnak. A populációdinamika célja annak megértése, hogy milyen tényezők befolyásolják egy populáció növekedését, csökkenését vagy földrajzi elterjedését.

A populációk változásait számos biológiai és környezeti tényező határozza meg. Az egyedek születése és halálozása, a bevándorlás és kivándorlás, a fajon belüli és fajok közötti interakciók mind fontos szerepet játszanak a populáció méretének alakulásában. A különböző növekedési modellek – például az exponenciális és logisztikus növekedés – segítenek a populációk dinamikájának leírásában és előrejelzésében. Egy adott élőhelyen a fajok közötti interakciók, úgymint zsákmányok és ragadozók, vagy éppen a versengés, egy komplex hálózatot alkotnak. A populációdinamika segít megérteni hogy ezek az interakciók hogyan befolyásolják az egyes fajok egyedszámának változását, az ökoszisztéma egészének dinamikáját és stabilitását.

A populációdinamika nemcsak az ökológiai rendszerek megértésében játszik kulcsszerepet, hanem számos gyakorlati alkalmazása is van. Az inváziós fajok terjedésének modellezése, a népesedéspolitika, a fenntartható halászat megtervezése, a fertőző betegségek terjedésének előrejelzése és a természetvédelmi stratégiák kidolgozása mind ezen tudományterület eredményein alapulnak.

A populációdinamika nemcsak ökológiai rendszerekre használható, hanem az orvostudományban is, hiszen a sejtjeink is populációkat alkotnak. Egy tumor növekedésének leírásához a rákos és egészséges sejtek versengését, az immunsejtek populációjával való kölcsönhatását, a tumor környezetének változását (például gyógyszeres beavatkozás révén) is figyelembe kell venni.
A mikrobiomunk is egy elképesztően gazdag, több ezer fajból álló ökoszisztéma, aminek jelentőségét és egészségünkre gyakorolt hatását még csak éppen kezdjük megérteni.

A populációdinamika körébe tartoznak a biológiai evolúciót leíró modellek is, például hogy egyes génváltozatok hogyan tudnak elterjedni. Néha egészen meglepő területeken is alkalmaznak populációdinamikai megközelítéseket, úgymint a hadászatban, a nyelvészetben, a pénzügyi szférában, a politológiában, vagy a kriminalisztikában.

A populációdinamika fejlődése szorosan összefonódik a matematikai modellezéssel. A dinamikai rendszerek, differenciálegyenletek,  mátrixmodellek és a számítógépes szimulációk lehetővé teszik a különböző populációs folyamatok pontosabb megértését és a hosszú távú trendek előrejelzését. ### Szinkronizáltan osztódó sejtpopuláció


$M_0, M_1, ..., M_n$, ahol $M_n$ az $n$-edik generációban a sejtek száma. \\
$M_{n+1}=a\cdot M_n$ \\
Mi történik a populációval hosszú távon? \\
$M_{n+1}=a^n\cdot M_0$, ahol $M_0$ a kezdeti populáció. \\
Ha $a=1$, akkor nem változik a populáció mérete, \\
ha $|a|>1 \Rightarrow |M_n|\rightarrow \infty$, azaz a populáció mérete nő, \\
ha $|a|<1 \Rightarrow |M_n|\rightarrow 0$, azaz a populáció mérete csökken

###Rovarpopuláció (levéltetű)


Teljes életciklust tekintünk:

$a_n$: felnőtt nőstény rovar az n-edik generációban \\
$p_n$: peték száma az n-edik generációban \\
$m$: halálozási arány/ráta a peték esetén amíg felnőtté válnak \\
$f$: egy nőstény által lerakott peték száma \\
$r$: nőstények aránya \\
$c$ reprodukciós szám \\
\begin{align*}
\left.
    \begin{array}{l}
        p_{n+1} &=  fa_n\\
        a_{n+1} &=  r(1-m)p_{n+1}
    \end{array}
\right\}
\end{align*}
amiből kapjuk, hogy:
\begin{align*}
a_{n+1} &=  rf(1-m)a_n,
\end{align*}
ahol $c=rf(1-m)$, tehát
\begin{align*}
a_{n+1} &=  ca_n
\end{align*}
Cél lehet: legyen $c<1$. Az $m$ változtatásával: $rf(1-m)<1$ átrendezve $1-m<\frac{1}{rf}$ vagyis $1-\frac{1}{rf}<m$

$p_n$: növények száma az $n$-edik évben \\

$s_n^2$: kétéves magok áprilisban \\
$\overset{\sim}{s}_n^1$: egyéves magok májusban \\
$\overset{\sim}{s}_n^2$: kétéves magok májusban \\
$s_n^0$: új magok augusztusban \\
\begin{split}
p_{n+1} & = \text{(egyévesből hajtott ki) + (kétévesből hajtott ki)} \\
 & = α\cdot s_n^1+β\cdot s_n^2
\end{split}
Emellett $\overset{\sim}{s}_n^1=(1-α)s_n^1$, ahol $s_{n+1}^1=σ\cdot s_n^0$, amiből következik, hogy $s_n^1=σ \cdot s_{n-1}^0$. \\
Hasonlóan $\overset{\sim}{s}_n^2=(1-β)s_n^2$, ahol  $s_{n+1}^2=σ\cdot \overset{\sim}{s}_n^1$, amiből következik, hogy $s_n^1=σ \cdot s_{n-1}^0$. \\
Emellett $s_n^0=γ \cdot p_n$.
Tehát a modell úgy fog alakulni, hogy \\
\begin{split}
p_{n+1} & = α \cdot s_{n+1}^1+β \cdot s_{n+1}^2= \\

\end{split}


###Lineáris differenciaegyenletek (diszkrét dinamikai rendszer)

## Nemlineáris diszkrét idejű modellek (egyfajos)
$x_{n+1}=f(x_n)$, ahol $f$ nemlineáris \\
(i.) $f(x)=rx(1-x)$ r>0 \\
(ii.) $f(x)=\frac{a⋅x}{b+x}$, $x_0, x_1, …, x_n$ \\


###1. Egyensúlyi helyzet:


$\overline{x}=f(\overline{x})$ egyenlet megoldásai $x_0=\overline{x}$, akkor $x_n = \overline{x}$, minden $n$-re \\

Példa: (ii.) \\
$\frac{a⋅\overline{x}}{b+\overline{x}}=\overline{x}$, ahol $a,b>0$, ekkor: \\
$\overline{x}=0$, és \\
$\frac{a}{b+\overline{x}}=1$, amiből kapjuk, hogy $\overline{x}=a-b$ \\

###2. Periodikus megoldás


$x_{m+p} = x_m$, és $x_{m+p} \neq x_{m+i}$, ahol $0<i<p$, minden $m$-re, akkor $x_0, x_1, x_2, …$ $p$-periodikus megoldás. \\
Példa: (ii.) van-e 2-periodikus megoldás? \\
\begin{align*}
\left.
    \begin{array}{l}
        f(f(x)) & = x \\
        f(x) & \neq x
    \end{array}
\right\} \Leftrightarrow \frac{a \frac{a \cdot x}{b+x}}{b + \frac{a \cdot x}{b+x}} = x, \quad \textrm{ahol }x  \neq 0
\end{align*}
$ \frac{a^2}{b+x} = b + \frac{a⋅x}{b+x} \Leftrightarrow \frac{\frac{a^2}{b+x}}{b + \frac{a\cdot x}{b+x}} = 1$ \\
$a^2=b(b+x)+a\cdot x$ \\
$a^2-b^2=(a+b)x$ \\
$a-b=x \Rightarrow$ nincs 2-periodikus megoldás \\

````python
import numpy as np
import matplotlib.pyplot as plt
import warnings

# A figyelmeztetések elnyomása
warnings.filterwarnings("ignore", category=UserWarning, module='matplotlib')

# Alapértelmezett betűtípus beállítása
plt.rcParams.update({'font.size': 16})  # Ne használj LaTeX-et

def plot_cobweb(f, r, x0, nmax=40):
    """Pókháló diagram rajzolása."""
    x = np.linspace(0, 1, 500)
    fig, ax = plt.subplots(figsize=(6, 4))

    # f(x) és y = x függvények
    ax.plot(x, f(x, r), c='#444444', lw=2)
    ax.plot(x, x, c='#444444', lw=2)

    # Iterációs lépések
    px, py = np.empty((2, nmax+1, 2))
    px[0], py[0] = x0, 0
    for n in range(1, nmax, 2):
        px[n] = px[n-1]
        py[n] = f(px[n-1], r)
        px[n+1] = py[n]
        py[n+1] = py[n]

    # Iteráció útvonalának megjelenítése
    ax.plot(px, py, c='b', alpha=0.7)

    # Tengelycímkék
    ax.set_xlabel('x')
    ax.set_ylabel('f(x)')
    ax.set_title(f'x_0 = {x0:.1f}, r = {r:.2f}')

    plt.show()

class AnnotatedFunction:
    """Matematikai függvény osztály."""
    def __init__(self, func):
        self.func = func

    def __call__(self, *args, **kwargs):
        return self.func(*args, **kwargs)

# Logisztikus leképezés f(x) = r * x * (1 - x)
func = AnnotatedFunction(lambda x, r: r * x * (1 - x))

# Diagramok rajzolása
plot_cobweb(func, 2.8, 0.2)
````

### Bifurkációs diagram


````python
import numpy as np
import matplotlib.pyplot as plt

def logistic_eq(r,x):
    return r*x*(1-x)

# Create the bifurcation diagram
def bifurcation_diagram(seed, n_skip, n_iter, step=0.0001, r_min=0):
    R = []
    X = []

    # Create the r values to loop. For each r value we will plot n_iter points
    r_range = np.linspace(r_min, 4, int(1/step))

    for r in r_range:
        x = seed;
        # For each r, iterate the logistic function and collect datapoint if n_skip iterations have occurred
        for i in range(n_iter+n_skip+1):
            if i >= n_skip:
                R.append(r)
                X.append(x)

            x = logistic_eq(r,x);
    # Plot the data
    plt.plot(R, X, ls='', marker=',')
    plt.ylim(0, 1)
    plt.xlim(r_min, 4)
    plt.xlabel('r')
    plt.ylabel('X')
    plt.show()

bifurcation_diagram(0.2, 100, 10)
bifurcation_diagram(0.2, 100, 10, r_min=2.8)
````

````python
import copy
import warnings

from ipywidgets import interact
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import odeint
from scipy.optimize import fsolve
import sympy

def f_eq(xs: np.ndarray, ts: np.ndarray, obj):
    return obj.f(x=xs)


class Modeller:
  def __init__(self, ps, t_max, x_0):
    self.ps = ps
    self.x_0 = x_0
    self.t_max = t_max
    self.t = np.linspace(0, t_max, 100)
    self.get_equilibria()

  def f(self, x):
    return self.ps["k"] * x * (self.ps["K"] - x) - self.ps["b"]

  def calc_roots(self, bs, sign: str):
    add_1 = self.ps["k"] * self.ps["K"]
    add_2 = np.sqrt((self.ps["k"] * self.ps["K"]) ** 2 - 4 * self.ps["k"] * bs)
    if sign == "+":
      return (add_1 + add_2) / (2 * self.ps["k"])
    else:
      return (add_1 - add_2) / (2 * self.ps["k"])

  def calc_peak_arg(self):
    return self.ps["k"] * self.ps["K"] ** 2 / 4

  def get_equilibria(self):
    p = sympy.Symbol('p')
    roots = sympy.solvers.solve(self.f(x=p), p)
    if roots[0].is_real:
      self.equilibria = roots
    else:
      self.equilibria = []

  def are_there_equilibria(self):
    return len(self.equilibria) > 0

  def get_valid_solution(self):
    with warnings.catch_warnings():
      warnings.simplefilter("ignore")
      sol = odeint(func=f_eq, y0=self.x_0, t=self.t,
                  args=(self, ))

    # negatív értékek kiszűrése
    if np.sum(sol < 0) > 0:
      thresh = np.where(sol < 0)[0][0]
      self.t = self.t[:thresh]
      sol = sol[:thresh]

    return sol

  def plot_direction_field(self, ax):
    t_m, x_m = np.meshgrid(np.linspace(0, self.t_max, 20),
                           np.linspace(0, self.ps["K"], 20))
    u_m = 1
    v_m = f_eq(xs=x_m, ts=t_m, obj=self)
    ax.quiver(t_m, x_m, u_m, v_m)

  def plot_equilibria(self, ax):
    if self.are_there_equilibria():
      ax.plot(self.t, self.equilibria[0] * np.ones(len(self.t)),
              c="red")
      ax.plot(self.t, self.equilibria[1] * np.ones(len(self.t)),
              c="darkgreen")

  def plot_solution(self, ax):
    sol = self.get_valid_solution()
    ax.plot(self.t, sol, c="darkblue")

  def plot_equilibria_b(self, ax):
    peak_arg = self.calc_peak_arg()
    b_plot_a = np.linspace(0, peak_arg, 100)

    roots_plot_1_a = self.calc_roots(bs=b_plot_a, sign="+")
    roots_plot_2_a = self.calc_roots(bs=b_plot_a, sign="-")

    ax.plot(b_plot_a, roots_plot_1_a, c="pink")
    ax.plot(b_plot_a, roots_plot_2_a, c="pink")

  def plot_actual_equilibria(self, ax):
    ax.axvline(self.ps["b"])
    if self.are_there_equilibria():
      ax.plot(self.ps["b"], self.calc_roots(bs=self.ps["b"], sign="+"),
              c="darkgreen", markersize=10, marker="o")
      ax.plot(self.ps["b"], self.calc_roots(bs=self.ps["b"], sign="-"),
              c="red", markersize=10, marker="o")

````

````python
from IPython.display import clear_output

@interact(k=(0, 1, 0.05),
          K=(0, 10, 0.5),
          b=(0, 5, 0.1),
          x_0=(0, 10, 1.0),
          t_max=(10, 30, 5.0))
def plot(k: float = 0.15, K: float = 10.0,
         b: float = 1.0, x_0: float = 5.0,
         t_max: float = 20):
  # Töröljük az előző ábrát
  clear_output(wait=True)

  # 0. Változó paraméterek alapján létrehozott változók
  ps_i = {"k": k, "K": K, "b": b}  # paraméter dict
  modeller = Modeller(ps=ps_i, t_max=t_max, x_0=x_0)

  # 1. Üres ábrák létrehozása
  fig, ax = plt.subplots(1, 2, figsize=(15, 5))
  # Néhány formázás
  fig.suptitle("Logisztikus szaporodás konstans vadászattal")
  ax[0].set_xlim(0, 5)
  ax[0].set_ylim(0, K)
  ax[1].set_ylim(0, K)
  ax[0].set_xlabel("b")
  ax[0].set_ylabel("egyensúlyi helyzet")
  ax[1].set_xlabel("t")
  ax[1].set_ylabel("populáció")

  # 2. Iránymező
  modeller.plot_direction_field(ax=ax[1])

  # 3. Egyensúlyi helyzetek
  modeller.plot_equilibria(ax=ax[1])

  # 4. Megoldás
  modeller.plot_solution(ax=ax[1])

  # 5. Egyensúlyi helyzetek diagramja
  modeller.plot_equilibria_b(ax=ax[0])

  # 6. Aktuális paraméterhez tartozó egyensúlyi helyzetek
  modeller.plot_actual_equilibria(ax=ax[0])

  # Ábra megjelenítése
  plt.show()
````
